I denne opgave skal I programmere et bibliotekssystem ved hjælp af objektorienteret programmering i Jupyter Notebook

a)  Lav en klasse Bog, som repræsenterer en bog med følgende attributter: titel, forfatter og udlånt (boolean)

b)  Tilføj følgende metoder
        - udlån(), som markerer bogen som udlånt
        - aflever(), som markerer bogen som afleveret
        - erUdlånt(), som returnerer, om bogen er udlånt
        - visInfo(), som udskriver information om bogen

c)  Lav en ny klasse Ebog, som arver fra klassen Bog, og derudover har oplysninger om filstørrelse (MB)

d)  Lav et program, der 
        - opretter tre normale bøger,
        - opretter to ebøger,
        - udskriver information om bøgerne og ebøgerne
        - udlåner en bog
        - tjekker, om den er udlånt

e)  Lav en klasse Bibliotek, der skal kunne
        - gemme flere bøger
        - udlåne bøger
        - vise alle bøger
        
f)  Lav et program, der tjekker, at biblioteket fungerer

g)  Udvid biblioteket med lydbøger

h)  Udvid med mulighed for køb af nogle af bøgerne

```python
class Bog:
    def __init__(self, titel, forfatter, udlånt=False, salgspris=None):
        self.titel = titel
        self.forfatter = forfatter
        self.udlånt = udlånt
        self.salgspris = salgspris
        self.solgt = False

    def udlån(self):
        if self.solgt:
            print(f"'{self.titel}' er solgt og kan ikke udlånes.")
            return
        if self.udlånt:
            print(f"'{self.titel}' er allerede udlånt.")
        else:
            self.udlånt = True
            print(f"'{self.titel}' er nu udlånt.")

    def aflever(self):
        if self.udlånt:
            self.udlånt = False
            print(f"'{self.titel}' er afleveret.")
        else:
            print(f"'{self.titel}' var ikke udlånt.")

    def erUdlånt(self):
        return self.udlånt

    def kanKøbes(self):
        return self.salgspris is not None and not self.solgt

    def køb(self):
        if self.kanKøbes():
            self.solgt = True
            self.udlånt = False
            print(f"'{self.titel}' er købt for {self.salgspris} kr.")
        elif self.solgt:
            print(f"'{self.titel}' er allerede solgt.")
        else:
            print(f"'{self.titel}' er ikke til salg.")

    def visInfo(self):
        status = "Udlånt" if self.udlånt else "Hjemme"
        salgsinfo = (
            f"Til salg ({self.salgspris} kr.)" if self.kanKøbes() else
            "Solgt" if self.solgt else
            "Ikke til salg"
        )
        print(f"Bog: {self.titel} | Forfatter: {self.forfatter} | Status: {status} | {salgsinfo}")


class Ebog(Bog):
    def __init__(self, titel, forfatter, filstørrelse_mb, udlånt=False, salgspris=None):
        super().__init__(titel, forfatter, udlånt, salgspris)
        self.filstørrelse_mb = filstørrelse_mb

    def visInfo(self):
        status = "Udlånt" if self.udlånt else "Hjemme"
        salgsinfo = (
            f"Til salg ({self.salgspris} kr.)" if self.kanKøbes() else
            "Solgt" if self.solgt else
            "Ikke til salg"
        )
        print(
            f"E-bog: {self.titel} | Forfatter: {self.forfatter} | "
            f"Filstørrelse: {self.filstørrelse_mb} MB | Status: {status} | {salgsinfo}"
        )


class Lydbog(Bog):
    def __init__(self, titel, forfatter, varighed_minutter, udlånt=False, salgspris=None):
        super().__init__(titel, forfatter, udlånt, salgspris)
        self.varighed_minutter = varighed_minutter

    def visInfo(self):
        status = "Udlånt" if self.udlånt else "Hjemme"
        salgsinfo = (
            f"Til salg ({self.salgspris} kr.)" if self.kanKøbes() else
            "Solgt" if self.solgt else
            "Ikke til salg"
        )
        print(
            f"Lydbog: {self.titel} | Forfatter: {self.forfatter} | "
            f"Varighed: {self.varighed_minutter} min | Status: {status} | {salgsinfo}"
        )


class Bibliotek:
    def __init__(self, navn):
        self.navn = navn
        self.bøger = []

    def tilføj_bog(self, bog):
        self.bøger.append(bog)

    def find_bog(self, titel):
        for bog in self.bøger:
            if bog.titel.lower() == titel.lower():
                return bog
        return None

    def udlån_bog(self, titel):
        bog = self.find_bog(titel)
        if bog:
            bog.udlån()
        else:
            print(f"Bogen '{titel}' blev ikke fundet.")

    def køb_bog(self, titel):
        bog = self.find_bog(titel)
        if bog:
            bog.køb()
        else:
            print(f"Bogen '{titel}' blev ikke fundet.")

    def vis_alle_bøger(self):
        print(f"\nBibliotek: {self.navn}")
        print("-" * 70)
        for bog in self.bøger:
            bog.visInfo()
        print("-" * 70)


# d) Opret tre normale bøger og to ebøger
bog1 = Bog("Den Lille Prins", "Antoine de Saint-Exupéry", salgspris=79)
bog2 = Bog("Hobbitten", "J.R.R. Tolkien")
bog3 = Bog("1984", "George Orwell", salgspris=99)
ebog1 = Ebog("Python for Begyndere", "A. Jensen", filstørrelse_mb=5.2, salgspris=149)
ebog2 = Ebog("Dataanalyse i Praksis", "M. Hansen", filstørrelse_mb=12.4)

print("\nInformation om bøger og ebøger:")
for b in [bog1, bog2, bog3, ebog1, ebog2]:
    b.visInfo()

# Udlån en bog og tjek om den er udlånt
print("\nUdlån og status-tjek:")
bog2.udlån()
print(f"Er '{bog2.titel}' udlånt?", bog2.erUdlånt())


# e, f, g, h) Bibliotek med bøger, ebøger, lydbøger og køb
bibliotek = Bibliotek("Byens Bibliotek")
lydbog1 = Lydbog("Sapiens (Audio)", "Yuval Noah Harari", varighed_minutter=900, salgspris=199)
lydbog2 = Lydbog("Harry Potter 1 (Audio)", "J.K. Rowling", varighed_minutter=510)

for item in [bog1, bog2, bog3, ebog1, ebog2, lydbog1, lydbog2]:
    bibliotek.tilføj_bog(item)

print("\nFuld oversigt i biblioteket:")
bibliotek.vis_alle_bøger()

print("\nTester udlån via bibliotek:")
bibliotek.udlån_bog("1984")
print("Er '1984' udlånt?", bibliotek.find_bog("1984").erUdlånt())

print("\nTester køb via bibliotek:")
bibliotek.køb_bog("Den Lille Prins")
bibliotek.køb_bog("Dataanalyse i Praksis")

print("\nOversigt efter udlån og køb:")
bibliotek.vis_alle_bøger()
```
