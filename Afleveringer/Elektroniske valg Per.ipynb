```python
import random

def hemmeligStemme(stemme):
    # genererer tre tal, der lagt sammen giver værdien af stemme, som er 0 eller 1
    a = random.randint(-10, 10)
    b = random.randint(-10, 10)
    c = stemme - a - b
    return a, b, c

class Person:
    def __init__(self, navn):
        self.navn = navn
        self.stemme = -1
        
    def stem(self):
        svar = ' '
        while svar not in 'jn':
            svar = input(f'{self.navn} - stemmer du ja eller nej - tast j eller n? ')
        if svar == 'j':
            self.stemme = 1
        else:
            self.stemme = 0
    
    def fordelStemme(self, serverA, serverB, serverC):           
        stemA, stemB, stemC = hemmeligStemme(self.stemme)
        serverA.modtag(stemA)
        serverB.modtag(stemB)
        serverC.modtag(stemC)

class Server:
    def __init__(self, navn):
        self.navn = navn
        self.delstemmer = []
        
    def modtag(self, delstemme):
        self.delstemmer.append(delstemme)
        
    def delsum(self):
        return sum(self.delstemmer)
    
personer = [Person('Anne'),
            Person('Bent'),
            Person('Clara'),
            Person('Dea'),
            Person('Esben'),
            Person('Frede'),
            Person('Gurli'),
            Person('Henriette'),
            Person('Ib')]

sA = Server('A')
sB = Server('B')
sC = Server('C')

for person in personer:
    person.stem()
    person.fordelStemme(sA, sB, sC)
    
jaSigere = sA.delsum() + sB.delsum() + sC.delsum()
nejSigere = len(personer) - jaSigere

print(f'\n{jaSigere} har stemt Ja \n{nejSigere} har stemt Nej')
```
