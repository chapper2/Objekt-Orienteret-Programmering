```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

methan = mpimg.imread(r'C:\Users\caspe\Documents\GitHub\Objekt-Orienteret-Programmering\Afleveringer\billeder til afl6\methan.png')
ethan = mpimg.imread(r'C:\Users\caspe\Documents\GitHub\Objekt-Orienteret-Programmering\Afleveringer\billeder til afl6\ethan.png')
propan = mpimg.imread(r'C:\Users\caspe\Documents\GitHub\Objekt-Orienteret-Programmering\Afleveringer\billeder til afl6\propan.png')
```

```python
import random
import time
from IPython.display import clear_output

alkaner = {'Methan':methan, 'Ethan':ethan, 'Propan':propan}
alkanNavne = ['Methan', 'Ethan', 'Propan']

rigtig = 0
forkert = 0
for i in range(len(alkanNavne)):
    quizStruktur = random.choice(alkanNavne)
    
    plt.imshow(alkaner[quizStruktur])
    plt.axis('off')
    plt.show()
            
    svar = input('hvilken Alkan ser du på billedet?')
        
    if svar == quizStruktur:
            print('Tillykke du har svaret rigtig')
            rigtig += 1
            
            clear_output(wait = True)
            
    else:
            print('Desværre, det er ikke det rigtige navn, prøv igen')
            forkert += 1
            clear_output(wait = True)
            
time.sleep(1)
print('-----------------------------------------------------------')
print(f'Du har {rigtig} antal rigtige, og {forkert} antal forkert')
print('-----------------------------------------------------------')

   
```
