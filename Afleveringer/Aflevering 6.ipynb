```python
import matplotlib.pyplot as plt
#import matplotlib.image as mpimg
import cv2

methan = cv2.imread("billeder til afl6/methan.jpg")
ethan = cv2.imread("billeder til afl6/ethan.jpg")
propan = cv2.imread("billeder til afl6/propan.jpg")
butan = cv2.imread("billeder til afl6/butan.jpg")
pentan = cv2.imread("billeder til afl6/pentan.jpg")
hexan = cv2.imread("billeder til afl6/hexan.jpg")
heptan = cv2.imread("billeder til afl6/heptan.jpg")
octan = cv2.imread("billeder til afl6/octan.jpg")
nonan = cv2.imread("billeder til afl6/nonan.jpg")
decan = cv2.imread("billeder til afl6/decan.jpg")

alkaner = {'Methan':methan, 'Ethan':ethan, 'Propan':propan, 'Butan':butan, 'Pentan':pentan, 'Hexan':hexan, 'Heptan':heptan, 'Octan':octan, 'Nonan':nonan, 'Decan':decan}
alkanNavne = ['Methan', 'Ethan', 'Propan', 'Butan', 'Pentan', 'Hexan', 'Heptan', 'Octan', 'Nonan', 'Decan']
```

```python
#version 1 - funktionel programmering

import random
import time
from IPython.display import clear_output

rigtig = 0
forkert = 0
for i in range(len(alkanNavne)):
    quizStruktur = random.choice(alkanNavne)
    
    plt.imshow(alkaner[quizStruktur])
    plt.axis('off')
    plt.show()
            
    svar = input('hvilken Alkan ser du på billedet?')
        
    if svar == quizStruktur:
            print('Tillykke du har svaret rigtig')
            rigtig += 1
            time.sleep(0.5)
            clear_output(wait = True)
            
    else:
            print('Desværre, det er ikke det rigtige navn, prøv igen')
            forkert += 1
            time.sleep(0.5)
            clear_output(wait = True)
            
time.sleep(1)
print('-----------------------------------------------------------')
print(f'Du har {rigtig} antal rigtige, og {forkert} antal forkert')
print('-----------------------------------------------------------')

   
```

```python
#version 2 - med klasser

import random
import time
from IPython.display import clear_output


class OrganiskQuiz():
    def __init__(self, antalAlkaner):
        self.antalAlkaner = antalAlkaner
        self.rigtig = 0
        self.forkert = 0

    def Game(self):
        alkan = Alkan()

        for i in range(self.antalAlkaner):
            navn = alkan.vælgTilfældigNavn()

            plt.imshow(alkan.vælgStruktur())
            plt.axis('off')
            plt.show()

            svar = input('Hvilken alkan ser du på billedet? ')

            # accepterer korrekt stavning uanset store/små bogstaver
            if svar.strip().lower() == navn.lower():
                print('Tillykke, du har svaret rigtigt')
                self.rigtig += 1
            else:
                print(f'Desværre, det rigtige svar er {navn}')
                self.forkert += 1

            time.sleep(0.8)
            clear_output(wait=True)

        print('-----------------------------------------------------------')
        print(f'Du har {self.rigtig} antal rigtige, og {self.forkert} antal forkerte')
        print('-----------------------------------------------------------')


class Alkan():
    def __init__(self):
       self.alkaner = {'Methan':methan, 'Ethan':ethan, 'Propan':propan, 'Butan':butan, 'Pentan':pentan, 'Hexan':hexan, 'Heptan':heptan, 'Octan':octan, 'Nonan':nonan, 'Decan':decan}
       self.alkanNavne = ['Methan', 'Ethan', 'Propan', 'Butan', 'Pentan', 'Hexan', 'Heptan', 'Octan', 'Nonan', 'Decan']
       self.quizStruktur = 'hej'

    def vælgTilfældigNavn(self):
        self.quizStruktur = random.choice(self.alkanNavne)
        return self.quizStruktur

    def vælgStruktur(self):
       return self.alkaner[self.quizStruktur]
```

```python
#test runde af celle 4

spil1 = OrganiskQuiz(10)
spil1.Game()
```

```python
# version 3 - udkast med mere encapsulation og tydelig metodeopdeling

import random
import time
from IPython.display import clear_output


class AlkanKatalogV2:
    def __init__(self):
        # Én samlet struktur, så quiz-klassen ikke skal kende de enkelte billedvariabler, og benytter nu key/value par til at sikre korrekt navn istedet for navneliste.
        self._alkaner = {
            'Methan': methan,
            'Ethan': ethan,
            'Propan': propan,
            'Butan': butan,
            'Pentan': pentan,
            'Hexan': hexan,
            'Heptan': heptan,
            'Octan': octan,
            'Nonan': nonan,
            'Decan': decan,
        }

    def hent_tilfældigt_spørgsmål(self):
        navn = random.choice(list(self._alkaner.keys()))
        return navn, self._alkaner[navn]


class OrganiskQuizV2:
    def __init__(self, antal_alkaner):
        self.antal_alkaner = antal_alkaner
        self.rigtig = 0
        self.forkert = 0
        self.katalog = AlkanKatalogV2()

    def game(self):
        for _ in range(self.antal_alkaner):
            self._spil_en_runde()
        self._vis_resultat()

    def _spil_en_runde(self):
        korrekt_navn, billede = self.katalog.hent_tilfældigt_spørgsmål()
        self._vis_billede(billede)

        svar = input('Hvilken alkan ser du på billedet? ')

        if self._er_korrekt_svar(svar, korrekt_navn):
            self.rigtig += 1
            print('Tillykke, du har svaret rigtigt')
        else:
            self.forkert += 1
            print(f'Desværre, det rigtige svar er {korrekt_navn}')

        # Lille pause før skærmen ryddes, så brugeren kan nå at læse feedback
        time.sleep(0.8)
        clear_output(wait=True)

    def _vis_billede(self, billede):
        plt.imshow(billede)
        plt.axis('off')
        plt.show()

    def _er_korrekt_svar(self, svar, korrekt_navn):
        # Sammenligner uden hensyn til store/små bogstaver og ekstra mellemrum
        return svar.strip().lower() == korrekt_navn.lower()

    def _vis_resultat(self):
        print('-----------------------------------------------------------')
        print(f'Du har {self.rigtig} antal rigtige, og {self.forkert} antal forkerte')
        print('-----------------------------------------------------------')
```

```python
# Eksempel på kørsel af det nye udkast
spil2 = OrganiskQuizV2(10)
spil2.game()
```
