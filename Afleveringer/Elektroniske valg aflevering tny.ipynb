I skal udvide programmet med elektroniske valg, så det kan håndtere valg, hvor der er flere partier, der kan stemmes på - f.eks. et folketingsvalg med fem partier A, B, C, D og E.
Bemærk, at algoritmen fortsat skal gøre brug af MPC.

I kan desuden - hvis der er tid til det - udvide programmet med partier som objekter, hvor I skal overveje tilhørende data og metoder.

Algoritmen gennemfører en anonym afstemning ved hjælp af secure multi-party computation (MPC) med tre servere.

Hver vælger afgiver sin stemme ved at vælge ét parti. Denne stemme oversættes til en numerisk værdi, hvor det valgte parti får værdien 1, og alle andre partier får værdien 0. For hvert parti opdeles værdien (0 eller 1) i tre tilfældige delstemmer, som tilsammen summerer til den oprindelige værdi.

De tre delstemmer sendes til hver sin server. Hver server gemmer kun sine egne delstemmer i en liste og har derfor ikke adgang til hele stemmen eller viden om, hvilket parti vælgeren har stemt på.

Efter alle vælgere har stemt, beregner hver server en delsum ved at lægge sine delstemmer sammen for hvert parti. Til sidst samles de tre serveres delsummer for et parti, hvilket giver det endelige stemmetal for dette parti.

```python
import random

def hemmeligStemme(stemme):
    # genererer tre tal, der lagt sammen giver værdien af stemme, som er 0 eller 1
    a = random.randint(-5,5)
    b = random.randint(-5,5)
    c = stemme-a-b
    return a, b, c
```

```python

class Person:
     # genererer tre tal, der lagt sammen giver værdien af stemme, som er 0 eller 1
    def __init__(self, navn):
        self.navn = navn
        self.stemme = -1
        
    def stem(self):
        svar = ' '
        while svar not in 'ABCDE':
            svar = input(f'{self.navn} - Stem på A, B, C, D eller E: ')
        self.stemme = svar 


#  fordeler stemmerne til de forskellige partier. 
# Hver person sender tre "hemmelige" stemmer til hver servers parti.
# Dvs summen af de tre hemmelige stemmer til et parti er 1, hvis personen har stemt på det parti, ellers 0.
#sA, sB, sC, sD, sE er lister med tre servere for hvert parti
    def fordelStemme(self, sA, sB, sC, sD, sE):

        if self.stemme == 'A': 
            stemA, stemB, stemC = hemmeligStemme(1)
        else:
            stemA, stemB, stemC = hemmeligStemme(0)

        sA[0].modtag(stemA)
        sA[1].modtag(stemB)
        sA[2].modtag(stemC)


        if self.stemme == 'B':
            stemA, stemB, stemC = hemmeligStemme(1)
        else:
            stemA, stemB, stemC = hemmeligStemme(0)

        sB[0].modtag(stemA)
        sB[1].modtag(stemB)
        sB[2].modtag(stemC)


        if self.stemme == 'C':
            stemA, stemB, stemC = hemmeligStemme(1)
        else:
            stemA, stemB, stemC = hemmeligStemme(0)

        sC[0].modtag(stemA)
        sC[1].modtag(stemB)
        sC[2].modtag(stemC)


        if self.stemme == 'D':
            stemA, stemB, stemC = hemmeligStemme(1)
        else:
            stemA, stemB, stemC = hemmeligStemme(0)

        sD[0].modtag(stemA)
        sD[1].modtag(stemB)
        sD[2].modtag(stemC)


        if self.stemme == 'E':
            stemA, stemB, stemC = hemmeligStemme(1)
        else:
            stemA, stemB, stemC = hemmeligStemme(0)

        sE[0].modtag(stemA)
        sE[1].modtag(stemB)
        sE[2].modtag(stemC)


        
```

```python
class Server:
    def __init__(self, navn):
        self.navn = navn
        self.delstemmer = []  # liste med alle delstemmer
        
    def modtag(self, delstemme):
        self.delstemmer.append(delstemme)  # tilføjer delstemme fra person
        
    def delsum(self):
        return sum(self.delstemmer)  # summen af alle elementer i listen destemmer
```

```python
personer = [Person('Anne'),
            Person('Bent'),
            Person('Clara'),
            Person('Dea'),
            Person('Esben'),
            Person('Frede'),
            Person('Gurli'),
            Person('Henriette'),
            Person('Ib')]


# Hvert parti har 3 servere (samme MPC-struktur som før)
# dvs. at der her er 3 server objekter for hvert parti
sA = [Server("A1"), Server("A2"), Server("A3")]
sB = [Server("B1"), Server("B2"), Server("B3")]
sC = [Server("C1"), Server("C2"), Server("C3")]
sD = [Server("D1"), Server("D2"), Server("D3")]
sE = [Server("E1"), Server("E2"), Server("E3")]


for person in personer:
    person.stem()   
    person.fordelStemme(sA, sB, sC, sD, sE)
    
# Udregn og print de endelige stemmetal for hvert parti
print("Parti A:", sA[0].delsum() + sA[1].delsum() + sA[2].delsum())
print("Parti B:", sB[0].delsum() + sB[1].delsum() + sB[2].delsum())
print("Parti C:", sC[0].delsum() + sC[1].delsum() + sC[2].delsum())
print("Parti D:", sD[0].delsum() + sD[1].delsum() + sD[2].delsum())
print("Parti E:", sE[0].delsum() + sE[1].delsum() + sE[2].delsum())


```
