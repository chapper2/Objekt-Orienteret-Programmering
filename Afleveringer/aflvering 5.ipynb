```python
#VERSION3
import random   #importere random for at kunne lave tilfældige tal

class Game():   #klasse for spillet
    def __init__(self, playerNumbers):
        self.__gameOver = False                     #spillet er slut ikke slut
        self.playerList=[]                          #liste til at indeholde spillere
        self.playerList = [Player(i+1) for i in range(playerNumbers)] #playerList fylde med spillere (List comprehension)


    def isGameOver(self): #funktion der tjekker om spillet er slut
        return self.__gameOver
    
    def checkPlayers(self):  #checker spillerne
        for player in self.playerList:
            if player.getHealth() <= 0:
                 # Udskriv dødsmeddelelse
                lovedOnes = ['Thomas', 'Mom', 'Lover']
                print(f'Spiller {player.number} dør - tell my {random.choice(lovedOnes)}')
                #self.playerList.pop(player.number-1) #smider spiller af listen
        
        #laver ny liste med de spillere der er i live (list comprehension)
        alive = [p for p in self.playerList if p.getHealth() > 0]
        if len(alive) <= 1: #hvis der kun er en levende spiller
            self.__gameOver = True #spiller slutter
            print(f"Spiller {alive[0].number} vinder!")

        #Status
        
        if self.__gameOver:
            print("spilelt er slut")
        else:
            print("spillet er i gang")

    
    def showPlayers(self):  #status for spillerne, der nu også viser våben skade
            for player in self.playerList:                                
                if player.getHealth() <= 0:
                    print(f'Spiller {player.number} er død')
                else:
                    print(f'Spiller {player.number} har {player.getHealth()} i liv. Dit våben skader {player.weaponDamage}')

           
           
class Player():   #klasse for spiller der nu også har et våben der laver tilfældig skade
    def __init__(self, number):
        self.number = number
        self.__health = 100           #Vi indkapler attributten health så den kun kan tilgås via player klassesn accessor metode
        self.weaponDamage = random.randint(1,100)    
     
    def takeDamage(self,damage):    #funktion der trækker skade fra spillerens liv
        self.__health -= damage 

    def getHealth(self):            #viser health for spiller med
        return self.__health      

    def attack(self):
        ready = False #holder styr på om spilleren har valgt et muligt offer
        while not ready:
            victim = int(input(f'Du er spiller {self.number}. Hvilket spillernummer vil du angribe? ')) 
        
            # Check om spillernummeret er gyldigt
            if victim < 1 or victim > len(game.playerList):
                print(f"Ugyldigt spillernummer! Vælg mellem 1 og {len(game.playerList)}")
            # Check om det er sig selv
            elif victim == self.number:
                print("Du kan ikke angribe dig selv!")
                # Check om målet er død
            elif game.playerList[victim-1].getHealth() <= 0:
                print(f'Spiller {victim} er allerede død!, stop med at forstyrre de døde')
            else:
                ready = True
    
        # Udfør angrebet
        game.playerList[victim-1].takeDamage(self.weaponDamage)
        print(f'Spiller {self.number} angriber spiller {victim} og laver {self.weaponDamage} i skade')

game = Game(3)
game.showPlayers()
while not game.isGameOver(): #så længe spillet ikke er slut
    #vælger en tilfældig spiller der er i live til at angribe  (list comprehension)
    attacker = random.choice([p for p in game.playerList if p.getHealth() > 0])    
    attacker.attack()
    game.showPlayers()    
    game.checkPlayers()
    





```
