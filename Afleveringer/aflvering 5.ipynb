```python
#VERSION3
import random

class Game():
    def __init__(self, playerNumbers):
        self.playerList=[]
        for i in range(playerNumbers):
            self.playerList.append(Player(i+1))
    
    def checkPlayers(self):
        for player in self.playerList:
            if player.health <= 0:
                self.playerList.pop(player.number-1)

    
    def showPlayers(self):
        for player in self.playerList:
            print(f'Spiller {player.number} har {player.health} i liv. Dit våben skader {player.weaponDamage}')

           
class Player():
    def __init__(self, number):
        self.number = number
        self.health = 100
        self.weaponDamage = random.randint(1,100)    
     
    def takeDamage(self,damage):
        self.health -= damage        

    def attack(self):
        victim = int(input("Hvilket spillernummer vil du angribe"))
        game.playerList[victim-1].takeDamage(self.weaponDamage)

    def dying(self):
        if self.health <= 0:
            lovedOnes = ['Thomas', 'Mom', 'Lover']
            print(f'dying, tell my {random.choice(lovedOnes)}')

game = Game(3)
game.showPlayers()
while len(game.playerList) != 1:
    game.playerList[random.randint(0,len(game.playerList)-1)].attack()
    game.checkPlayers()



```
