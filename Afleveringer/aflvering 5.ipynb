```python
#VERSION2
#Der kan opretts et spil med et vilkårligt antal spillere, og hver spiller har et våben der laver tilfældig skade på en anden spiller

import random   #importere random for at kunne lave tilfældige tal

class Game():   #klasse for spillet
    def __init__(self, playerNumbers):
        self.playerList=[]                          #liste til at indeholde spillere
        for i in range(playerNumbers):              #Der laves en liste med spillere
            self.playerList.append(Player(i+1))

        
    
    def showPlayers(self):  #status for spillerne, der nu også viser våben skade
        for player in self.playerList:
            print(f'Spiller {player.number} har {player.getHealth()} i liv. Dit våben skader {player.weaponDamage}')

           
class Player():   #klasse for spiller der nu også har et våben der laver tilfældig skade
    def __init__(self, number):
        self.number = number
        self.__health = 100           #Vi indkapler attributten health så den kun kan tilgås via player klassesn accessor metode
        self.weaponDamage = random.randint(1,100)    
     
    def takeDamage(self,damage):    #funktion der trækker skade fra spillerens liv
        self.__health -= damage 

    def getHealth(self):            #viser health for spiller med
        return self.__health      

    def attack(self):   #funktion der gør det muligt for spilleren at angribe en anden spiller
        victim = int(input("Hvilket spillernummer vil du angribe")) #input for at vælge hvilken spiller man vi angribe
        game.playerList[victim-1].takeDamage(self.weaponDamage)

game = Game(3)              #der laves et spil med 3 spillere
game.showPlayers()          #status for spillerne vises
game.playerList[0].attack() #spiller 1 angriber en anden spiller
game.showPlayers()          #status for spillerne vises 


```
