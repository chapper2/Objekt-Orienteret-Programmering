```python
import random

class Player:
    def __init__(self, number):
        self.number = number
        self._health = 100
        self.weapon_damage = random.randint(1, 100)
    
    def __str__(self):
        return f"Spiller {self.number} (HP: {self._health})"

    @property
    def is_alive(self):
        return self._health > 0

    def take_damage(self, damage):
        if not self.is_alive:
            return

        self._health -= damage
        if self._health <= 0:
            self._health = 0
            print(f"Spiller {self.number} er død")

    def attack(self, target):
        if not self._validate_attack(target):
            return

        print(
            f"Spiller {self.number} angriber spiller {target.number} "
            f"og laver {self.weapon_damage} skade"
        )
        target.take_damage(self.weapon_damage)

    def _validate_attack(self, target):
        if not self.is_alive:
            print(f"Spiller {self.number} er død og kan ikke angribe")
            return False

        if not target.is_alive:
            print(f"Spiller {target.number} er allerede død")
            return False

        return True

class Mage(Player):
    def __init__(self, number):
        super().__init__(number)
        self.weapon_damage = 70
        self._mana = 50
        self._mana_cost = 10
    
    def __str__(self):
        return f"Mage {self.number} (HP: {self._health}, Mana: {self._mana})"

    def attack(self, target):
        if not self._validate_attack(target):
            return

        if self._mana < self._mana_cost:
            print(f"Mage {self.number} har ikke nok mana")
            return

        self._mana -= self._mana_cost
        print(f"Mage {self.number} kaster en ildkugle på spiller {target.number} "
            f"og laver {self.weapon_damage} skade")
        
        target.take_damage(self.weapon_damage)

class Game:
    def __init__(self):
        self.players = self._create_players()

    def _create_players(self):
        return ([Player(i + 1) for i in range(3)]
            + [Mage(i + 4) for i in range(3)])

    def _alive_players(self):
        return [player for player in self.players if player.is_alive]

    def _alive_players_count(self):
        return len(self._alive_players())

    def _get_winner(self):
        alive = self._alive_players()
        return alive[0] if len(alive) == 1 else None

    def play(self):
        while self._alive_players_count() > 1:
            attacker, target = random.sample(self._alive_players(), 2) 
            attacker.attack(target)

        winner = self._get_winner()
        print("Vinderen er:")
        print(winner)
        
```
