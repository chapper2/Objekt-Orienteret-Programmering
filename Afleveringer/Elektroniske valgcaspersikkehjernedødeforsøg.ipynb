```python
import random

def hemmeligStemme(stemme):
    """
    Opdeler en stemme i tre tilfældige delstemmer, der tilsammen giver den originale stemmeværdi.
    Dette sikrer anonymitet gennem Multi-Party Computation.
    """
    # Genererer tre tal, der lagt sammen giver værdien af stemme, som er 0 eller 1
    a = random.randint(-5,5)  # Første tilfældige delstemme
    b = random.randint(-5,5)  # Anden tilfældige delstemme
    c = stemme-a-b  # Tredje delstemme beregnes så summen bliver korrekt
    return a, b, c
```

```python
class Person:
    """
    Klasse der repræsenterer en stemmeberettiget person.
    Personen kan afgive sin stemme og fordele den til tre servere.
    """
    def __init__(self, navn):
        """Initialiserer en person med navn og tom stemme."""
        self.navn = navn
        self.stemme = []
        
    def stem(self):
        """
        Lader personen afgive sin stemme ved at vælge x, y eller z.
        Stemmen gemmes som en liste med tre værdier [x, y, z].
        """
        svar = ' '
        while svar not in 'xyz':
            svar = input(f'{self.navn} - stemmer du x, y eller z, indtast bogstav og tryk enter ')
        # Konverterer svaret til en liste med 1 for det valgte alternativ og 0 for de andre
        if svar ==  'x':
            self.stemme = [1,0,0]
        elif svar ==  'y':
            self.stemme = [0,1,0]
        elif svar ==  'z':
            self.stemme = [0,0,1]
    
    def fordelStemme(self, serverA, serverB, serverC):
        """
        Fordeler personens stemme til tre servere ved at opdele den i hemmelige delstemmer.
        Dette sikrer at ingen enkelt server kan se den faktiske stemme.
        """           
        # Udpakker stemmeværdierne for x, y og z
        x,y,z = self.stemme
        
        # Opdeler hver stemmeværdi i tre hemmelige delstemmer
        ax, bx, cx = hemmeligStemme(x)
        ay, by, cy = hemmeligStemme(y)
        az, bz, cz = hemmeligStemme(z)
       
        # Sender delstemmerne til de tre servere
        serverA.modtagX(ax); serverB.modtagX(bx); serverC.modtagX(cx)
        serverA.modtagY(ay); serverB.modtagY(by); serverC.modtagY(cy)
        serverA.modtagZ(az); serverB.modtagZ(bz); serverC.modtagZ(cz)
```

```python
class Server:
    """
    Klasse der repræsenterer en server i Multi-Party Computation systemet.
    Serveren opsamler delstemmer og beregner delsummer uden at kende de faktiske stemmer.
    """
    def __init__(self, navn):
        """Initialiserer serveren med et navn og tomme lister til delstemmer."""
        self.navn = navn
        self.delstemmerX = []  # Liste med alle delstemmer for alternativ X
        self.delstemmerY = []  # Liste med alle delstemmer for alternativ Y
        self.delstemmerZ = []  # Liste med alle delstemmer for alternativ Z
        
    def modtagX(self, delstemme):
        """Modtager og gemmer en delstemme for alternativ X."""
        self.delstemmerX.append(delstemme)  # Tilføjer delstemme fra person
        
    def modtagY(self, delstemme):
        """Modtager og gemmer en delstemme for alternativ Y."""
        self.delstemmerY.append(delstemme) 
    
    def modtagZ(self, delstemme):
        """Modtager og gemmer en delstemme for alternativ Z."""
        self.delstemmerZ.append(delstemme)     
        
    def delsumX(self):
        """Beregner og returnerer delsummen for alle X-stemmer på denne server."""
        return sum(self.delstemmerX)
    
    def delsumY(self):
        """Beregner og returnerer delsummen for alle Y-stemmer på denne server."""
        return sum(self.delstemmerY)
    
    def delsumZ(self):
        """Beregner og returnerer delsummen for alle Z-stemmer på denne server."""
        return sum(self.delstemmerZ)
```

```python
# Opretter en liste af personer med deres navne
personer = [Person('Anne'), Person('Bent'),
            Person('Clara'),
            Person('Dea'),
            Person('Esben'),
            Person('Frede'),
            Person('Gurli'),
            Person('Henriette'),
            Person('Ib')]

# Initialiserer servere A, B og C
sA = Server('A')
sB = Server('B')
sC = Server('C')

# Itererer gennem hver person i listen
for person in personer:
    person.stem()  # Kalder stemme-funktionen for hver person
    person.fordelStemme(sA, sB, sC)  # Fordeler stemmen til de tre servere

# Udskriver resultatet af stemmefordelingen
print("Resultat:")
print("X:", sA.delsumX() + sB.delsumX() + sC.delsumX())
print("Y:", sA.delsumY() + sB.delsumY() + sC.delsumY())
print("Z:", sA.delsumZ() + sB.delsumZ() + sC.delsumZ())
```
