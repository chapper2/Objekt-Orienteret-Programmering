# Elektroniske valg

```python
import random

def hemmeligStemme(stemme):
    # genererer tre tal, der lagt sammen giver værdien af stemme, som er 0 eller 1
    a = random.randint(-5,5)
    b = random.randint(-5,5)
    c = stemme-a-b
    return a, b, c
```

```python
class Person:
    def __init__(self, navn):
        self.navn = navn
        self.stemme = []
        
    def stem(self):
        svar = ' '
        while svar not in 'xyz':
            svar = input(f'{self.navn} - stemmer du x, y eller z, indtast bogstav og tryk enter ')
        if svar ==  'x':
            self.stemme = [1,0,0]
        elif svar ==  'y':
            self.stemme = [0,1,0]
        elif svar ==  'z':
            self.stemme = [0,0,1]
    
    def fordelStemme(self, serverA, serverB, serverC):           
        
        x,y,z = self.stemme
        
        ax, bx, cx = hemmeligStemme(x)
        ay, by, cy = hemmeligStemme(y)
        az, bz, cz = hemmeligStemme(z)
       
        serverA.modtagX(ax); serverB.modtagX(bx); serverC.modtagX(cx)
        serverA.modtagY(ay); serverB.modtagY(by); serverC.modtagY(cy)
        serverA.modtagZ(az); serverB.modtagZ(bz); serverC.modtagZ(cz)
        
        
            
```

```python
class Server:
    def __init__(self, navn):
        self.navn = navn
        self.delstemmerX = []  # liste med alle delstemmer
        self.delstemmerY = []
        self.delstemmerZ = []
        
        
    def modtagX(self, delstemme):
        self.delstemmerX.append(delstemme)  # tilføjer delstemme fra person
        
    def modtagY(self, delstemme):
        self.delstemmerY.append(delstemme) 
    
    def modtagZ(self, delstemme):
        self.delstemmerZ.append(delstemme)     
        
    def delsumX(self):
        return sum(self.delstemmerX)
    
    def delsumY(self):
        return sum(self.delstemmerY)
    
    def delsumZ(self):
        return sum(self.delstemmerZ)
        
```

```python
personer = [Person('Anne'), Person('Bent'),
            Person('Clara'),
            Person('Dea'),
            Person('Esben'),
            Person('Frede'),
            Person('Gurli'),
            Person('Henriette'),
            Person('Ib')]

sA = Server('A')
sB = Server('B')
sC = Server('C')

for person in personer:
    person.stem()  # skriv kommentar
    person.fordelStemme(sA, sB, sC)  # skriv kommentar
    

print("Resultat:")
print("X:", sA.delsumX() + sB.delsumX() + sC.delsumX())
print("Y:", sA.delsumY() + sB.delsumY() + sC.delsumY())
print("Z:", sA.delsumZ() + sB.delsumZ() + sC.delsumZ())


```
