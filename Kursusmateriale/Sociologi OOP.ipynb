```python
import random

class Person:
    def __init__(self, holdning, personlighed):
        self.holdning = holdning
        self.personlighed = personlighed
        
    def overtal(self, andenPerson):
        if andenPerson.personlighed == 'fleksibel' and andenPerson.holdning != self.holdning:
            andenPerson.holdning = self.holdning
            
class Befolkning:
    def __init__(self, antal):
        self.personer = []
        for _ in range(antal):
            holdning = random.choice(['ja', 'nej'])
            personlighed = random.choice(['fleksibel', 'stædig'])
            self.personer.append(Person(holdning, personlighed))
            
    def mingleParvist(self):
        random.shuffle(self.personer)

        for i in range(0, len(self.personer), 2):
            if i+1 < len(self.personer):
                p1 = self.personer[i]
                p2 = self.personer[i+1]

                p1.overtal(p2)
                
    def antalHoldninger(self):
        jaSigere = 0
        for p in self.personer:
            if p.holdning == 'ja':
                jaSigere += 1
        nejSigere = len(self.personer) - jaSigere
        return jaSigere, nejSigere
    
```

```python
befolkning = Befolkning(antal=100)

print(f'Startfordeling: {befolkning.antalHoldninger()}')

for runde in range(30):
    befolkning.mingleParvist()
    print(f'Efter {runde+1}. runde: {befolkning.antalHoldninger()}')
    
```
