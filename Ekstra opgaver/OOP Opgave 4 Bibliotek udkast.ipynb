```python
#   a)  Lav en klasse Bog, som repræsenterer en bog med følgende attributter: titel, forfatter og udlaant (boolean)

#   b)  Tilføj følgende metoder
#       - udlaan(), som markerer bogen som udlånt
#       - aflever(), som markerer bogen som afleveret
#      - erUdlaant(), som returnerer, om bogen er udlånt
#       - visInfo(), som udskriver information om bogen

class Book:
    def __init__(self, tittel, forfatter):
        self.tittel = tittel
        self.forfatter = forfatter
        self.udlaant = False
     
    def erUdlaant(self):
        return self.udlaant
  
    def visInfo(self):
        if self.erUdlaant():
            status = "Er udlånt"
        else:
            status = "Er ledig"
        return f"{self.tittel} af {self.forfatter} ({status})"

    def udlaan(self):
        if self.erUdlaant() == True:
            return f"{self.tittel} er allerede udlånt."
        else:
            self.udlaant = True
            return f"{self.tittel} er nu udlånt."

    def aflever(self):
        if self.erUdlaant() == True:
            self.udlaant = False
            return f"{self.tittel} er nu returneret."
        else:
            return f"{self.tittel} var ikke udlånt."
```

```python
#   c)  Lav en ny klasse Ebog, som arver fra klassen Bog, og derudover har oplysninger om filstørrelse (MB)


class eBook(Book):
    def __init__(self, tittel, forfatter, filstørrelse):
        super().__init__(tittel, forfatter,)
        self.filstørrelse = filstørrelse
```

```python

#   d)  Lav et program der:

# Opretter tre normale bøger
bog1 = Book("1984", "George Orwell")
bog2 = Book("Harry Potter", "J.K. Rowling")
bog3 = Book("Sapiens", "Yuval Noah Harari")

# Opretter to eBøger
ebog1 = eBook("Clean Code", "Robert C. Martin", 5.4)
ebog2 = eBook("Python Crash Course", "Eric Matthes", 6.1)

# Samler alle bøger i en liste
bibliotek = [bog1, bog2, bog3, ebog1, ebog2]

# Udskriver information om alle bøger og eBøger
print("Bøger i biblioteket:")
for bog in bibliotek:
    print(bog.visInfo())

print()

# Udlåner en bog
print(bog1.udlaan())

# Tjekker om den er udlånt
print( bog1.visInfo())
```

```python
#   e)  Lav en klasse Bibliotek, der skal kunne

        # gemme flere bøger
class Bibliotek:
        def __init__(self):
            self.boeger = [] # liste til at gemme bøger i biblioteket

        def tilfoejBog(self, tittel, forfatter):
            bog = Book(tittel, forfatter)
            self.boeger.append(bog)

        def tilfoejEBog(self, tittel, forfatter, filstoerrelse):
            ebog = eBook(tittel, forfatter, filstoerrelse)
            self.boeger.append(ebog)
        
       #def tilfoejLydBog(self, tittel, forfatter, filstoerrelse):
       #    lydbog = eBook(tittel, forfatter, filstoerrelse)
       #    self.boeger.append(lydbog)

        # vise alle bøger
        def visAlleBoeger(self):
            for bog in self.boeger:
                print(bog.visInfo()) # viser alle bøger i biblioteket ved polymorf brug af visInfo() metoden

        # udlåne bøger
        def udlaanBog(self, tittel):  # udlåner en bog baseret på titel, hvis den findes i biblioteket
            for bog in self.boeger:
                if bog.tittel == tittel:
                    return bog.udlaan()
            return f"Bog med titel '{tittel}' ikke fundet."
       
```

```python
#   f)  Lav et program, der tjekker, at biblioteket fungerer
bibliotek = Bibliotek()
bibliotek.tilfoejBog("1984", "George Orwell")
bibliotek.tilfoejBog("Harry Potter", "J.K. Rowling")
bibliotek.tilfoejEBog("On the Use and Abuse of History for Life", "Friedrich Nietzsche", 1.4)
bibliotek.tilfoejEBog( "The Liskov Substitution Principle", "Norvell, T.S", 0.19)
bibliotek.udlaanBog("1984") # Udlåner en bog for at tjekke at det virker
bibliotek.udlaanBog("Sapiens") # Grænsetilfælde prøver at udlåne en bog der ikke findes
bibliotek.udlaanBog("The Liskov Substitution Principle") # Grænsetilfælde Udlaan en Ebog for at tjekke at digitale bøger hentes i stet for at udlånes
bibliotek.visAlleBoeger()
```
