```python
#   a)  Lav en klasse Bog, som repræsenterer en bog med følgende attributter: titel, forfatter og udlaant (boolean)

#   b)  Tilføj følgende metoder
#       - udlaan(), som markerer bogen som udlånt
#       - aflever(), som markerer bogen som afleveret
#      - erUdlaant(), som returnerer, om bogen er udlånt
#       - visInfo(), som udskriver information om bogen

class Book:
    def __init__(self, titel, forfatter, antal):
        self._titel = titel
        self._forfatter = forfatter
        self._antal = antal
        self._udlaant = 0 # Privat attribut for at forhindre direkte adgang
     
    def erUdlaant(self):
        return self._udlaant
  
    def visInfo(self):
        print(f"{self._titel} af {self._forfatter}, {(self._antal-self._udlaant)} Styk tilgængelige")

    def udlaan(self):
        if self._udlaant < self._antal:
            self._udlaant += 1
        else:
            self._udlaant == self.antal
            print(f"{self._titel} er desværre ikke tilgængelig. Alle eksemplarer er i øjeblikket udlånt.")

    def aflever(self):
        if self._udlaant == 0:
            print(f"{self._titel} var ikke udlånt.")
        else:
            self._udlaant -= 1
            print(f"{self._titel} er nu returneret.")

#   h)  Udvid med mulighed for køb af nogle af bøgerne (Køb og salg)
    def koeb(self, antal):
        self._antal += antal
        print(f"{antal} eksemplarer af {self._titel} er nu tilføjet. Total antal: {self._antal}")
   
    def saelg(self, antal):
        if antal > self._antal - self._udlaant:
            print(f"Kan ikke sælge {antal} eksemplarer af {self._titel}. Kun {self._antal - self._udlaant} tilgængelige.")
        else:
            self._antal -= antal
            print(f"{antal} eksemplarer af {self._titel} er nu solgt. Total antal: {self._antal}") 
```

```python
#   c)  Lav en ny klasse Ebog, som arver fra klassen Bog, og derudover har oplysninger om filstørrelse (MB)

from typing import override


class eBook(Book):
    def __init__(self, titel, forfatter, antal, filstørrelse):
        super().__init__(titel, forfatter, antal)
        self._filstørrelse = filstørrelse
    
    @override
    def visInfo(self): # Override visInfo metoden for at inkludere filstørrelse og antal gange e-bogen er blevet hentet
        super().visInfo() # Kalder den oprindelige visInfo metode for at vise titel og forfatter
        print(f"Filen fylder {self._filstørrelse} MB")
    
```

```python

#   d)  Lav et program der:

# Opretter tre normale bøger
bog1 = Book("1984", "George Orwell",1)
bog2 = Book("Harry Potter", "J.K. Rowling",2)
bog3 = Book("Sapiens", "Yuval Noah Harari",3)

# Opretter to eBøger
ebog1 = eBook("Clean Code", "Robert C. Martin",200, 5.4)
ebog2 = eBook("Python Crash Course", "Eric Matthes",100, 6.1)

# Samler alle bøger i en liste
bibliotek = [bog1, bog2, bog3, ebog1, ebog2]

# Udskriver information om alle bøger og eBøger
print("Bøger i biblioteket:")
for bog in bibliotek:
    bog.visInfo()

print()

# Udlåner en bog
bog1.udlaan()

#Udlåner en eBog for at tjekke at den tæller antal gange den er blevet hentet i stedet for at markere den som udlånt
ebog1.udlaan()
ebog1.udlaan()

# Tjekker om den er udlånt
bog1.visInfo()
ebog1.visInfo()
```

```python

#   e)  Udvid klassen Ebog til også at kunne repræsentere lydbøger, ved at tilføje varighed (i timer) og oplæser som attributter.
#       Override visInfo() metoden for at inkludere disse oplysninger.
#   g)  Udvid biblioteket med lydbøger
class audioBook(eBook):
    def __init__(self, titel, forfatter, antal, filstørrelse,varighed, narrator):
        super().__init__(titel, forfatter, antal, filstørrelse) # Kalder super().__init__ for at initialisere titel, forfatter og filstørrelse fra eBook klassen
        self._varighed = varighed
        self._narrator = narrator

    @override
    def visInfo(self):
        super().visInfo()
        print(f"Varighed {self._varighed} Oplæser: {self._narrator}")
```

```python
#   e)  Lav en klasse Bibliotek, der skal kunne

        # gemme flere bøger
class Bibliotek:
        def __init__(self):
            self._materialer = [] # liste til at gemme bøger i biblioteket

        def tilfoejMateriale(self, materiale):
            self._materialer.append(materiale)

        # vise alle bøger
        def visAlleMaterialer(self):
            for materiale in self._materialer:
                materiale.visInfo() # viser alle bøger i biblioteket ved polymorf brug af visInfo() metoden

```

```python
#   f)  Lav et program, der tjekker, at biblioteket fungerer
bibliotek = Bibliotek()
bibliotek.tilfoejMateriale(Book("1984", "George Orwell", 1))
bibliotek.tilfoejMateriale(Book("Harry Potter", "J.K. Rowling", 1))
bibliotek.tilfoejMateriale(eBook("On the Use and Abuse of History for Life", "Friedrich Nietzsche", 100, 1.4))
bibliotek.tilfoejMateriale(eBook("The Liskov Substitution Principle", "Norvell, T.S", 200, 0.19))
bibliotek.tilfoejMateriale(audioBook("The Power of Habit", "Charles Duhigg", 300, 10, 150, "John Lee"))
#bibliotek.bog.udlaan("1984") # Udlåner en bog for at tjekke at det virker
# bibliotek.udlaanMaterialer("Sapiens") # Grænsetilfælde prøver at udlåne en bog der ikke findes
# bibliotek.udlaanMaterialer("The Liskov Substitution Principle") # Grænsetilfælde Udlaan en Ebog for at tjekke at digitale bøger hentes i stet for at udlånes
bibliotek.visAlleMaterialer()
```
