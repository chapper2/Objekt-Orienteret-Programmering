```python
class Person:
    def __init__(self, navn, alder):
        self.navn = navn
        self.alder = alder
        
    def hils(self):
        print('Hey, jeg hedder ' + self.navn + ', og jeg er ' + str(self.alder) + ' år gammel')
    
    def foedselsdag(self):
        self.alder += 1
        
    def skiftNavn(self):
        nytNavn = input('Hvad ønsker du et hedde, ' + self.navn + ' ? :')
        self.navn = nytNavn

person1 = Person('Simon', 37)
person1.hils()
person1.foedselsdag()
person1.hils()
person1.skiftNavn()
person1.hils()
```

## Sange

```python
class Sang:
    def __init__(self, artist, titel):
        self.artist = artist
        self.titel = titel
        
    def spiller(self):
        return 'Spiller: ' + self.titel + ' af ' + self.artist
    
sang1 = Sang('Zar Paulo', 'Klap For Fædrelandet')
sang1.spiller()

sang2 = Sang('ABBA', 'Mamma Mia')
sang2.spiller()

sang3 = Sang('Creedence Clearwater Revival', 'Have You Ever Seen the Rain?')
sang3.spiller()
```

```python
class ToDoListe:
    def __init__(self, opgave):
        self.opgave = opgave
        self.udfoert = False
        
    def done(self):
        self.udfoert = True
        return self.opgave + ' er udført'

jobs = []
jobs.append(ToDoListe('Vaske op'))
jobs.append(ToDoListe('Støvsuge'))
jobs.append(ToDoListe('Spise frokost'))
jobs.append(ToDoListe('Hvile sig'))
jobs.append(ToDoListe('Rydde sne'))

jobs[2].done()
jobs[3].done()

for job in jobs:
    if job.udfoert:
        print(job.opgave + ': Er udført')
    else:
        print(job.opgave + ': Er endnu ikke udført')
```
