```python
import random

def hemmeligStemme(stemme):
    # genererer tre tal, der lagt sammen giver værdien af stemme, som er 0 eller 1
    a = random.randint(-5,5)
    b = random.randint(-5,5)
    c = stemme-a-b
    return a, b, c
```

```python
class Person:
    def __init__(self, navn):
        self.navn = navn
        self.stemme = -1
        
    def stem(self):
        svar = ' '
        while svar not in 'jn':
            svar = input(f'{self.navn} - stemmer du ja eller nej - tast j eller n? ')
        if svar ==  'j':
            self.stemme = 1  # skriv kommentar
        else:
            self.stemme = 0  # skriv kommentar
    
    def fordelStemme(self, serverA, serverB, serverC):           
        stemA, stemB, stemC = hemmeligStemme(self.stemme)  # skriv kommentar
        serverA.modtag(stemA)
        serverB.modtag(stemB)
        serverC.modtag(stemC)
```

```python
class Server:
    def __init__(self, navn):
        self.navn = navn
        self.delstemmer = []  # liste med alle delstemmer
        
    def modtag(self, delstemme):
        self.delstemmer.append(delstemme)  # tilføjer delstemme fra person
        
    def delsum(self):
        return sum(self.delstemmer)  # summen af alle elementer i listen destemmer
```

```python
personer = [Person('Anne'),
            Person('Bent'),
            Person('Clara'),
            Person('Dea'),
            Person('Esben'),
            Person('Frede'),
            Person('Gurli'),
            Person('Henriette'),
            Person('Ib')]

sA = Server('A')
sB = Server('B')
sC = Server('C')

for person in personer:
    person.stem()  # skriv kommentar
    person.fordelStemme(sA, sB, sC)  # skriv kommentar
    
print(f"Der er {sA.delsum()+sB.delsum()+sC.delsum()} ud af {len(personer)} vælgere der har stemt for.")


```
