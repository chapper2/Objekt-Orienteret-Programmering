```python
import random

class Terning():
    def __init__(self):
        self.antalTerninger = int(input('indtast antallet af terninger du vil slå med'))
        self.antalØjne = int(input('indtast antal øjne terningerne skal have'))
        self.øjeListe = []
        
        
        
    def terningeKast(self):
        for i in range(self.antalTerninger):
            resultat = random.randint(1,self.antalØjne)
            self.øjeListe.append(resultat)
        
        print(f'Du har slået med terningerne {self.antalTerninger} gange, og fået følgende øjne ud: {self.øjeListe}, summen af øjnene er {sum(self.øjeListe)}')
            
```

```python
terning1 = Terning()

terning1.terningeKast()
```
