```python

from datetime import date
import datetime

class Person:
    def __init__(self, navn, cpr):
        self.navn = navn
        self.__cpr = cpr
        

    def bestemAlder(self):
        dag = int(self.__cpr[:2])
        måned = int(self.__cpr[2:4])
        år = int(self.__cpr[4:6])
        
        nu = int(str(date.today().year)[2:])
        
        if år > nu:
            år += 1900
        else:
            år +=2000
        
        print(date(år, måned, dag))
        
        
        
        
        
```

```python
p1 = Person('Casper', '1508891357')

p1.bestemAlder()
```
