```python
class Elev:
    def __init__(self, navn):
        self.navn = navn
        self.karakter = -5
```

```python
class Hold:
    def __init__(self, holdliste):
        self.holdliste = []
        
    def nyElev(self):
        self.elev1 = Elev('Jens')
        self.elev2 = Elev('Jytte')
        self.elev3 = Elev('Jeppe')
        
    def nytHold(self):
        self.holdliste.append(self.elev1)
        self.holdliste.append(self.elev2)
        self.holdliste.append(self.elev3)
            
    def givKarakterer(self):
        self.elev1.karakter = 2
        self.elev2.karakter = 7
        self.elev3.karakter = 12
            
    def gennemsnit(self):
        karaktersum = self.elev1.karakter + self.elev2.karakter + self.elev3.karakter
        return karaktersum/len(self.holdliste)    
    
    def udskriv(self):
        print(f'{self.elev1.navn} har fået karakteren {self.elev1.karakter} og {self.elev2.navn} har fået karakteren {self.elev2.karakter} og {self.elev3.navn} har fået karakteren {self.elev3.karakter}')
                
```

```python
hold = Hold('1a dansk')
hold.nyElev()
hold.nytHold()
hold.givKarakterer()
hold.udskriv()
hold.gennemsnit()
```
