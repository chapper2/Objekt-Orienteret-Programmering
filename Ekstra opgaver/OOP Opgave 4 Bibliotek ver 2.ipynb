```python

from abc import ABC, abstractmethod

# Opretter en abstrakt klasse Materiale, som både Bog, eBog og Lydbog arver fra
class Materiale(ABC):
    def __init__(self, titel, forfatter, biblioteksNr, idNr=None):
        self._titel = titel
        self._forfatter = forfatter
        self._biblioteksNr = biblioteksNr
        self._idNr = idNr #ISBN/eISBN/ISSN/ISRC (valgfrit)

    @abstractmethod
    def getInfo(self):
        return (f"{self._titel} af {self._forfatter}, Biblioteksnummer: {self._biblioteksNr}")
    
    #Printer information om materialet afhængig af hvilken type det er (bog, eBog eller lydbog)
    def visInfo(self):
        print(self.getInfo())
       
```

```python
#   a)  Lav en klasse Bog, som repræsenterer en bog med følgende attributter: titel, forfatter og udlaant (boolean)

#   b)  Tilføj følgende metoder
#       - udlaan(), som markerer bogen som udlånt
#       - aflever(), som markerer bogen som afleveret
#      - erUdlaant(), som returnerer, om bogen er udlånt
#       - visInfo(), som udskriver information om bogen
from typing import override

class Book(Materiale):
    def __init__(self, titel, forfatter, biblioteksNr, idNr, udlaant=False):
        super().__init__(titel, forfatter, biblioteksNr, idNr)
        self._udlaant = udlaant
        
    def udlaan(self):
        if self._udlaant == False:
            self._udlaant = True
            print(f"{self._titel} er nu udlånt.")
        else:
            print(f"{self._titel} Dette eksemplar er i øjeblikket allerede udlånt.")

    def aflever(self):
        if self._udlaant == False:
            print(f"{self._titel} var ikke udlånt.")
        else:
            self._udlaant = False
            print(f"{self._titel} er nu returneret.")

    def erUdlaant(self):
        return self._udlaant
    
    @override
    def getInfo(self):
        return (
            super().getInfo()
            + " | "
            + (f"ISBN: {self._idNr} | " if self._idNr is not None else "")
            + ("Er udlånt" if self._udlaant else "Er ikke udlånt"))
```

```python
#   c)  Lav en ny klasse Ebog, som arver fra klassen Bog, og derudover har oplysninger om filstørrelse (MB)

from typing import override

class eBook(Materiale):
    def __init__(self, titel, forfatter, biblioteksNr, idNr, filstoerrelse=0, tilgaaet=0):
        super().__init__(titel, forfatter, biblioteksNr, idNr)
        self._filstoerrelse = filstoerrelse
        self._downloads = tilgaaet
    
    def udlaan(self):
        self._downloads += 1
        print(f"E-bogen '{self._titel}' er blevet hentet. Antal downloads: {self._downloads}")
    
    
    @override
    def getInfo(self):
        return (
            super().getInfo()
            + " | "
            + (f"eISBN: {self._idNr} | " if self._idNr is not None else "")
            + f"Filen fylder {self._filstoerrelse} MB | "
            + f"Antal downloads: {self._downloads}")

    
```

```python

#   d)  Lav et program der:

# Opretter to normale bøger
bog1 = Book("1984", "George Orwell",1, "978-0451524935")
bog2 = Book("Harry Potter", "J.K. Rowling",2, "978-0439701835")

# Opretter fire eBøger
ebog1 = eBook("Clean Code", "Robert C. Martin",4, "978-0134494166",5.4)
ebog2 = eBook("Python Crash Course", "Eric Matthes",5, "978-1593279288",6.1)
ebog3 = eBook("On the Use and Abuse of History for Life", "Friedrich Nietzsche",6,None,7.2)
ebog4 = eBook("The Pragmatic Programmer", "Andrew Hunt and David Thomas",7, "978-0201616224",8.3)

# Samler alle bøger i en liste
bibliotek = [bog1, bog2, ebog1, ebog2, ebog3, ebog4]

# Udskriver information om alle bøger og eBøger
print("Materialer i biblioteket:")
for materiale in bibliotek:
    materiale.visInfo()

print()

# Udlåner en bog
bog1.udlaan()

#Udlåner en eBog for at tjekke at den tæller antal gange den er blevet hentet i stedet for at markere den som udlånt
ebog1.udlaan()

# Tjekker om de er udlånt / hentet
bog1.visInfo()
ebog1.visInfo()

#Tjekker at en bog ikke kan udlånes igen, når den allerede er udlånt og at den kan afleveres
bog1.udlaan()
bog1.aflever()
bog1.visInfo()
```

```python

#   e)  Udvid klassen Ebog til også at kunne repræsentere lydbøger, ved at tilføje varighed (i timer) og oplæser som attributter.
#       Override visInfo() metoden for at inkludere disse oplysninger.
#   g)  Udvid biblioteket med lydbøger
from typing import override

# Opretter subklassen AudioBook, som arver indirekte fra Materialer (positional arguments) 
# og indirekte fra materialer (keyword arguments) via eBook, og som repræsenterer en lydbog
# og tilføjer varighed og oplæser som attributter
class AudioBook(eBook):
    def __init__(self, titel, forfatter, biblioteksNr, idNr, filstoerrelse=0, tilgaaet=0, varighed=0, oplaeser=""):
        super().__init__(titel, forfatter, biblioteksNr, idNr, filstoerrelse=filstoerrelse, tilgaaet=tilgaaet)
        self._varighed = varighed
        self._oplaeser = oplaeser

    def udlaan(self):
            self._downloads += 1
            print(f"Lydbogen '{self._titel}' er blevet hentet. Antal downloads: {self._downloads}")
    
    
    @override
    def getInfo(self):
            return (
            super().getInfo()
            + (f" | ISRC: {self._idNr}" if self._idNr is not None else "")
            + f" | Varighed: {self._varighed} timer"
            + f" | Oplæser: {self._oplaeser}")

```
