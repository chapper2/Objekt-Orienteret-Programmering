```python

from abc import ABC, abstractmethod

# Opretter en abstrakt klasse Materiale, som både Bog, eBog og Lydbog arver fra
class Materiale(ABC):
    def __init__(self, tittel, forfatter, biblioteksNr, standardId=None):
        self._tittel = tittel
        self._forfatter = forfatter
        self._biblioteksNr = biblioteksNr
        self._standardId = standardId

    def getTittel(self):
        return self._tittel

    def getForfatter(self):
        return self._forfatter
    
    def getBiblioteksNr(self):
        return self._biblioteksNr

    @abstractmethod
    def getStandardId(self):
        pass

    @abstractmethod
    def erDigital(self):
        pass

    @abstractmethod
    def getMaterialeType(self):
        pass

    @abstractmethod
    def getInfo(self):
        return (
            f"{self.getMaterialeType()}"
            f"{self._tittel} af {self._forfatter}, "
            f"Biblioteksnummer: {self._biblioteksNr}")

    #Printer information om materialet afhængig af hvilken type det er (bog, eBog eller lydbog)
    def visInfo(self):
        print(self.getInfo())
       
```

```python
#   a)  Lav en klasse Bog, som repræsenterer en bog med følgende attributter: titel, forfatter og udlaant (boolean)

#   b)  Tilføj følgende metoder
#       - udlaan(), som markerer bogen som udlånt
#       - aflever(), som markerer bogen som afleveret
#      - erUdlaant(), som returnerer, om bogen er udlånt
#       - visInfo(), som udskriver information om bogen
from typing import override

class Book(Materiale):
    def __init__(self, tittel, forfatter, biblioteksNr, standardId=None, udlaant=False):
        super().__init__(tittel, forfatter, biblioteksNr, standardId)
        self._udlaant = udlaant
    
    def udlaan(self):
        if self._udlaant == False:
            self._udlaant = True
            print(f"{self._tittel} er nu udlånt.")
        else:
            print(f"{self._tittel} Dette eksemplar er i øjeblikket allerede udlånt.")

    def aflever(self):
        if self._udlaant == False:
            print(f"{self._tittel} var ikke udlånt.")
        else:
            self._udlaant = False
            print(f"{self._tittel} er nu returneret.")

    def erUdlaant(self):
        return self._udlaant 
    
    @override
    def getStandardId(self):
        return self._standardId

    @override
    def erDigital(self):
        return False

    @override
    def getMaterialeType(self):
            return "    Bog: "

    @override
    def getInfo(self):
        return (super().getInfo()
            + (f", ISBN: {self._standardId}" if self._standardId is not None else "")
            + (" Er udlånt" if self._udlaant else " Er ikke udlånt"))
```

```python
#   c)  Lav en ny klasse Ebog, som arver fra klassen Bog, og derudover har oplysninger om filstørrelse (MB)

from typing import override

class eBook(Materiale):
    def __init__(self, tittel, forfatter, biblioteksNr, standardId=None, filstoerrelse=0, tilgaaet=0):
        super().__init__(tittel, forfatter, biblioteksNr, standardId)
        self._eISBN = standardId
        self._filstoerrelse = filstoerrelse
        self._downloads = tilgaaet
    
    def udlaan(self):
        self._downloads += 1
        print(f"e-bogen '{self._tittel}' er blevet hentet. Antal downloads: {self._downloads}")

    @override
    def getStandardId(self):
        return self._eISBN

    @override
    def erDigital(self):
            return True
    
    @override
    def getMaterialeType(self):
            return "  e-Bog: "
     
    @override
    def getInfo(self):
        return (super().getInfo()
            + (f", e-ISBN: {self._eISBN}" if self._eISBN is not None else "")
            + f" Filen fylder {self._filstoerrelse} MB "
            + f"Antal downloads: {self._downloads}")

    
```

```python

#   e)  Udvid klassen Ebog til også at kunne repræsentere lydbøger, ved at tilføje varighed (i timer) og oplæser som attributter.
#       Override visInfo() metoden for at inkludere disse oplysninger.
#   g)  Udvid biblioteket med lydbøger
from typing import override

# Opretter subklassen AudioBook, som arver indirekte fra Materialer (positional arguments) 
# og indirekte fra materialer (keyword arguments) via eBook, og som repræsenterer en lydbog
# og tilføjer varighed og oplæser som attributter
class AudioBook(eBook):
    def __init__(self, tittel, forfatter, biblioteksNr, standardId=None, filstoerrelse=0, tilgaaet=0, varighed=0, oplaeser=""):
        super().__init__(tittel, forfatter, biblioteksNr, standardId, filstoerrelse=filstoerrelse, tilgaaet=tilgaaet)
        self._varighed = varighed
        self._oplaeser = oplaeser

    def udlaan(self):
            self._downloads += 1
            print(f"Lydbogen '{self._tittel}' er blevet hentet. Antal downloads: {self._downloads}")
     
     # Acessor metode til biblioteksklassen for at tjekke antal downloads   
    def getDownloads(self):
        return self._downloads
    
    @override
    def getMaterialeType(self):
            return "e-Lydbog: "
    @override
    def getStandardId(self):
        return self._standardId
       
    @override
    def getInfo(self):
            return (super().getInfo()
            + f" | Varighed: {self._varighed} timer"
            + f" | Oplæser: {self._oplaeser}")

```

```python
#   e)  Lav en klasse Bibliotek, der skal kunne

        # gemme flere bøger
from matplotlib.artist import get


class Bibliotek:
        def __init__(self):
            self._materialer = [] # liste til at gemme bøger i biblioteket
            
        # Validering af materiale i forhold til eksisterende materialer i biblioteket for at sikre at det opfylder kravene før det tilføjes
        def _validering(self, materiale):
            eksisterende_numre = {m.getBiblioteksNr() for m in self._materialer}

            for eksisterende in self._materialer:
                if (eksisterende.getTittel() == materiale.getTittel() and
                    eksisterende.getForfatter() == materiale.getForfatter()):

                    # STANDARD-ID skal matche, hvis begge har et
                    if (eksisterende.getStandardId() is not None and 
                        materiale.getStandardId() is not None and 
                        eksisterende.getStandardId() != materiale.getStandardId()):
                        print("Fejl: Standard (ISBN/eISBN/ISRC) matcher ikke "
                            "eksisterende materiale med samme titel og forfatter.")
                        return None
                    
                    # Digitale materialer må kun eksistere én gang
                    if eksisterende.erDigital() and materiale.erDigital():
                        print("Fejl: Digitale materialer kan kun tilføjes én gang.")
                        return None

            # Find / tildel et gyldigt biblioteksnummer
            if materiale.getBiblioteksNr() in eksisterende_numre:
                return max(eksisterende_numre) + 1 if eksisterende_numre else 1

            return materiale.getBiblioteksNr()

        def tilfoejMateriale(self, materiale):
            # Valider materiale før det tilføjes
            valid_biblioteksNr = self._validering(materiale)
            if valid_biblioteksNr is None:
                print("Materiale kunne ikke tilføjes.")
                return

            # Biblioteksnummer skal være unikt
            string = ""
            if valid_biblioteksNr != materiale.getBiblioteksNr():
                string = f" (Biblioteksnummer ændret til {valid_biblioteksNr} for at undgå konflikt)"

            self._materialer.append(materiale)
            print(f"{materiale.getTittel()} er tilføjet biblioteket.{string}")
       
        # vis alle materialer i biblioteket
        def visAlleMaterialer(self):
            print("Bibliotekets materialer:")
            print()
            for materiale in self._materialer:
                materiale.visInfo() # viser alle bøger i biblioteket ved polymorf brug af visInfo() metoden

        def afleverMateriale(self, biblioteksNr):
            for materiale in self._materialer:
                if materiale.erDigital():
                    print("Fejl: Digitale materialer kan ikke afleveres.")
                    return
                if materiale.getBiblioteksNr() == biblioteksNr:
                    materiale.aflever()
                    return
            print("Materiale ikke fundet")

        def udlaanMateriale(self, biblioteksNr):
            for materiale in self._materialer:
                if materiale.getBiblioteksNr() == biblioteksNr:
                    materiale.udlaan()
                    return
            print("Materiale ikke fundet")

                #   h)  Udvid med mulighed for køb af nogle af bøgerne (Køb og salg)
      
        def koebMateriale(self, materiale):
            # Validering af materiale i forhold til eksisterende materialer i biblioteket for at sikre at det opfylder kravene før det tilføjes
            valid_biblioteksNr = self._validering(materiale)
            if valid_biblioteksNr is None:
                print("Materialet kunne ikke Købes.")
                return
            
            # Kun ét materiale købes
            self.tilfoejMateriale(materiale)
            print(f"{materiale.getTittel()} er blevet købt og tilføjet biblioteket.")
      
        def saelgBog(self, Tittel, Forfatter):
            beholdning = []

            # Liste med alle fysiske bøger, der matcher titel og forfatter
            for bog in self._materialer:
                if (bog.getTittel() == Tittel and
                    bog.getForfatter() == Forfatter and
                    not bog.erDigital()):
                    beholdning.append(bog)

            if not beholdning:
                print("Ingen fysiske eksemplarer af denne bog findes i biblioteket.")
                return

            # Sælg den første tilgængelige kopi, der ikke er udlånt
            for bog in beholdning:
                if not bog.erUdlaant():
                    streng = bog.getInfo()
                    self._materialer.remove(bog)
                    print(f"Biblioteket har solgt {streng}")
                    print(f"Der er nu {len(beholdning) - 1} fysiskeeksemplarer tilbage i biblioteket.")
                    return
            print(f"Der findes {len(beholdning)} eksemplarer af {Tittel}, men de er alle i øjeblikket udlånt og kan derfor ikke sælges.")

            
   
        



```

```python
#   f)  Lav et program, der tjekker, at biblioteket fungerer
bibliotek = Bibliotek()
bibliotek.tilfoejMateriale(Book("1984", "George Orwell", 1))
bibliotek.tilfoejMateriale(Book("1984", "George Orwell", 2)) # Tjekker at det er muligt at tilføje flere fysiske eksemplarer af samme bog
bibliotek.tilfoejMateriale(Book("1984", "George Orwell", 1)) # Tjekker at det er muligt at tilføje flere fysiske eksemplarer af samme bog og at de får unikke biblioteksnumre
bibliotek.tilfoejMateriale(Book("Harry Potter", "J.K. Rowling", 4))
bibliotek.tilfoejMateriale(eBook("On the Use and Abuse of History for Life", "Friedrich Nietzsche",2,None, 1.4))
bibliotek.tilfoejMateriale(eBook("The Liskov Substitution Principle", "Norvell, T.S",2,None, 0.19))
bibliotek.tilfoejMateriale(AudioBook("Clean Code", "Robert C. Martin",11, None,5.4, varighed=3, oplaeser="John Doe"))
bibliotek.tilfoejMateriale(AudioBook("Clean Code", "Robert C. Martin",12, None,5.4, varighed=3, oplaeser="Jane Doe"))

bibliotek.udlaanMateriale(1)  # Udlåner materiale nr 1 for at tjekke at det virker
bibliotek.udlaanMateriale(1)  # Prøver at udlåne materiale nr 1 igen for at tjekke at det ikke kan udlånes igen
bibliotek.udlaanMateriale(2)  # Udlåner materiale nr 2 for at tjekke at det virker
bibliotek.afleverMateriale(1)  # Afleverer materiale nr 1 for at tjekke at det virker
bibliotek.udlaanMateriale(6)  # Udlåner materiale nr 5 for at tjekke at det virker
bibliotek.afleverMateriale(6)  # Prøver at aflevere et digitalt materiale for at tjekke at det ikke kan afleveres
bibliotek.udlaanMateriale(77) # Grænsetilfælde prøver at udlåne en bog der ikke findes
bibliotek.afleverMateriale(77) # Grænsetilfælde prøver at aflevere en bog der ikke findes
bibliotek.udlaanMateriale(1) # Udlåner materiale nr 1 igen for at tjekke at det virker efter aflevering

bibliotek.visAlleMaterialer()
```
